#Write csv to Bronze Layer

###crm_cust_info.csv

```python
df = (
    spark.read.option("header","true")
    .option("inferSchema","true")
    .csv(["/Volumes/workspace/bronze/souce_data/source_crm/cust_info.csv"]
        )
    )
df.write.mode("overwrite").saveAsTable("bronze.crm_cust_info")
```
